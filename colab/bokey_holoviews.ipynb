```python
import pandas as pd
import numpy as np
import holoviews as hv
import panel as pn
from holoviews import opts

# 1. Initialize Bokeh backend for HoloViews
hv.extension('bokeh')
pn.extension()

# 2. Generate Sample Data
np.random.seed(42)
dates = pd.date_range("2023-01-01", periods=100)
df = pd.DataFrame({
    'Date': np.tile(dates, 2),
    'Price': np.random.randn(200).cumsum() + 100,
    'Stock': ['AAPL'] * 100 + ['GOOG'] * 100
})

# 3. Create HoloViews Curve Element
curve = hv.Curve(df, kdims=['Date'], vdims=['Price'], label='Stock Price')

# 4. Customize Interactivity with Bokeh Options
# Using hv.opts to add tools like hover, zoom, and pan
curve.opts(
    opts.Curve(
        width=800, height=400, tools=['hover', 'wheel_zoom'],
        title="Interactive Stock Explorer", color='blue', line_width=2
    )
)

# 5. Create a dynamic dashboard with Panel widgets
def get_stock(stock_name):
    # Filter data based on selection
    filtered_df = df[df['Stock'] == stock_name]
    return hv.Curve(filtered_df, kdims=['Date'], vdims=['Price']).opts(
        title=f"{stock_name} Price", color='red'
    )

# Create selector widget
stock_selector = pn.widgets.Select(name='Stock', options=['AAPL', 'GOOG'])

# Link widget to the function
interactive_plot = pn.bind(get_stock, stock_selector)

# 6. Display the app
app = pn.Column(stock_selector, interactive_plot)
app.servable()
```
